Question 49 : Write a python program to find whether the given number is perfect number or not
Perfect Number : The sum of the factors of the numbers is equal to the number

```python
n=int(input("Enter the number"))
if(n<=0):
    print("Invalid Input")
else:
    s=0
    for i in range(1,n):
        if(n%i==0):
            s=s+i
    if(s==n):
        print("Perfect Number")
    else:
        print("Not Perfect Number")
```

Question 63 : Write a python program to print the report of the student marks

```python
#Student Id number validation
while(True):
    sno=int(input())
    if(sno>0):
        break
    print("Invalid")
#Name validation
name=input("Enter your name")
if(len(name)==0):
    print("Try again")

#Marks
while(True):
    cm=int(input("Enter c marks"))
    if(0<=cm<=100):
        break
    print("Invalid")

while(True):
    cppm=int(input("Enter cpp marks"))
    if(0<=cppm<=100):
        break
    print("Invalid")

while(True):
    osm=int(input("Enter os marks"))
    if(0<=osm<=100):
        break
    print("Invalid")

#total Marks 
totmarks=cm+cppm+osm
percent=round((totmarks/300)*100,2)

if((cm<40) or (cppm<40) or (osm<40)):
    grade="Fail"
else:
    if((totmarks<=300) and (totmarks>=250)):
        grade="DISTINCTION"
    elif(250>totmarks>=200):
        grade="FIRST"
    elif(totmarks in range(150,200)):
        grade="SECOND"
    elif(totmarks in range(120,150)):
        grade="THIRD"

print("="*50)
print("STUDENT MARKS REPORT")
print("="*50)
print("Student Number : ",sno)
print("Student Name : ",name)
print("Marks")
print("C = ",cm)
print("CPP = ",cppm)
print("OS = ",osm)
print("-"*50)
print("Total Marks = ",totmarks)
print("Percentage = ",percent)
print("Grade = ",grade)
print("="*50)
```
