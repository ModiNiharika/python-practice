MODULES

The purpose of modules concept is that to reuse the functions , global variables and class names from one program to another program provided both the programs in same folder

There are 2 types of modules 

1) Predefined modules 

2) Programmer or user defined modules

Example of predefined Module

```python
import calendar
year = 2006
month = 11
print(calendar.month(year,month))
```

To reuse the features of module we have 2 approaches . They are 

> By using import statement

> By using from....import statement

By using import statement

```python
#Syntax 1 : import module name

import icici
print(icici.bname)
```

```python
#Syntax 2 : import module name1,module name2,.....module name n

import icici , MathsInfo
print("This is imported from icici file",icici.bname)
print("This is imported from MathsInfo file",MathsInfo.PI)
```

```python
#Syntax 3 : import module name as alias name

import icici as i
print(i.bname)
print(i.addr)
i.simpleint()
```

```python
#Syntax 4 : import module name1 as alias name1 , module name2 as alias name2......module name n as alias name n

import icici as i, MathsInfo as m
print("The PI value is ",m.PI)
print("The bank name is ",i.bname)
```
