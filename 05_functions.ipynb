Question 68 : Write a python program which will accept any value and decide its type

```python
def decidetype(value):
    if(type(value)==int):
        print("int type")
    elif(type(value)==float):
        print("float type")
    elif(type(value==bool)):
        print("bool type")
    elif(type(value)==complex):
        print("complex type")
    elif(type(value)==str):
        print("str type")
    elif(type(value)==tuple):
        print("tuple type")
    elif(type(value)==list):
        print("list type")
    elif(type(value)==bytes):
        print("bytes type")
    elif(type(value)==bytearray):
        print("bytearray type")
    elif(type(value)==set):
        print("set type")
    elif(type(value)==dict):
        print("dict type")
    else:
        print("None Type")
decidetype(True)
decidetype(12)
decidetype('niha')
```

Question 87 : Write a python program which will give following 

Line : PyT5on 12s aN Oo3p la9nn1G

Filter all upper case letters and concat them 

Filter all numerical values and concat them

Filter all special symbols and display them

```python
line = input("Enter the line: ")
# Uppercase letters
upper = ''.join(filter(lambda ch: ch.isupper(), line))
# Numerical values
numbers = ''.join(filter(lambda ch: ch.isdigit(), line))
# Special symbols
special = list(filter(lambda ch: not ch.isalnum() and not ch.isspace(), line))
print("Uppercase letters:", upper)
print("Numbers:", numbers)
print("Special symbols:", special)
```

Question 88 : Write a python program to implement the following
