INPUT AND OUTPUT FUNCTIONS

Input function

```python
print(50*'*')
print("Syntax 1: print(val1)  or print(val1,val2,....,valn)")
print(50*'-')
a=12
b=2
print(a)
print(a,b)
print(50*'*')

print("Syntax 2 : print(msg1) or print(msg1+msg2+...+msgn)")
print(50*'-')
print("Hello World")
print("Modi "+"Niharika")
print(50*'*')

print("Syntax 3 : print(msgs cum values or values cum msgs)")
print(50*'-')
print("Value of a =",a)
print(a,"is not b")
print(50*'*')


print("Syntax 4 : print(msgs cum values with format)")
print(50*'-')
print("Value of a = {}".format(a))
print("{} + {} = {} ".format(a,b,a+b))
print(50*'*')

print("Syntax 5 : print(messages cum values with format specifiers)")
print(50*'-')
print("Value of a ,b = %f,%f"%(a,b))
print(50*'*')

print("Syntax 6 : print(value,end=Delimeter)")
print(50*'-')
for val in range(1,4):
    print(val,end="--")
```

Question 2 : Write a python program which will accept length and breadth and calculate area of rectangle and its perimeter

```python
l=float(input("Enter length"))
b=float(input("Enter breadth"))
ra=l*b
pr=2*(l+b)
print("Area = ",ra)
print("Perimeter = ",pr)
```

Question 5 : Write a python program which will calculate simple interest and total amount to pay

```python
p=float(input("Enter principle amount"))
t=float(input("Enter time"))
r=float(input("Enter rate of interest"))
si=(p*t*r)/100
print("The interest is ",si)
print("The total amount is ",si+p)
```
